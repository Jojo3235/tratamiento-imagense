```python
import cv2, matplotlib.pyplot as plt, numpy as np

img = cv2.imread("image1.png", 0)
plt.imshow(img)
```

### Filtro de media

```python
n,m = np.shape(img)
n,m
```

```python
new_image = np.zeros((n-2,m-2))
plt.imshow(new_image, cmap= "gray")
```

```python
kernel_size = 15
lados = int((kernel_size-1)/2)


K = [1/kernel_size**2]*kernel_size**2
 
for i in range(1, n-lados):
    for j in range(1, m-lados):
        frag = []
        for x in range(-lados, lados+1):
            for y in range(-lados,lados+1):
                frag.append(img[i-y, j-x])
        new_image[i-1,j-1] = np.dot(K,frag)
        
```

```python
plt.imshow(new_image, cmap = "gray")
```
